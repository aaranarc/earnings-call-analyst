```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=150)
```

```python
doc = fitz.open("data/US/JPMorgan/JPMorgan_US_2024_Q1.pdf")

full_text = ""
for page in doc:
    full_text += page.get_text()

print(f"Total characters: {len(full_text)}")
```

```python
chunks = text_splitter.split_text(full_text)
print(f"Number of chunks: {len(chunks)}")
```

```python
print("--- Chunk 0 ---")
print(chunks[0])
print("\n--- Chunk 1 ---")
print(chunks[1])
```

```python
import os

def parse_filename(filepath):
    filename = os.path.basename(filepath)  # e.g. "JPMorgan_US_2024_Q1.pdf"
    name = filename.replace(".pdf", "")
    parts = name.split("_")  # ["JPMorgan", "US", "2024", "Q1"]
    return {
        "company": parts[0],
        "market": parts[1],
        "year": int(parts[2]),
        "quarter": parts[3]
    }

# test it
metadata = parse_filename("data/US/JPMorgan/JPMorgan_US_2024_Q1.pdf")
print(metadata)
```

```python
filepath = "data/US/JPMorgan/JPMorgan_US_2024_Q1.pdf"
metadata = parse_filename(filepath)

# Build a list of (chunk_text, metadata) pairs
documents_with_metadata = []
for chunk in chunks:
    chunk_metadata = metadata.copy()  # copy so each chunk gets its own dict, not a shared reference
    chunk_metadata["section"] = "unclassified"  # placeholder, since we're skipping section detection for now
    documents_with_metadata.append({
        "text": chunk,
        "metadata": chunk_metadata
    })

print(f"Total documents ready: {len(documents_with_metadata)}")
print(documents_with_metadata[0])
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

```python
%pip install chromadb
```

```python
import chromadb

client = chromadb.PersistentClient(path="./chroma_db")
collection = client.get_or_create_collection(name="earnings_calls")
```

```python
texts = [doc["text"] for doc in documents_with_metadata]
metadatas = [doc["metadata"] for doc in documents_with_metadata]
ids = [f"{metadata['company']}_{metadata['quarter']}_{metadata['year']}_chunk{i}" for i, metadata in enumerate(metadatas)]

# Generate embeddings for all chunks
embedded_vectors = embeddings.embed_documents(texts)

# Add everything to the collection
collection.add(
    documents=texts,
    embeddings=embedded_vectors,
    metadatas=metadatas,
    ids=ids
)

print(f"Added {collection.count()} documents to the collection")
```

```python
# Parse HDFC Q1
filepath_hdfc = "data/India/HDFC/HDFC_India_2024_Q1.pdf"
doc_hdfc = fitz.open(filepath_hdfc)

full_text_hdfc = ""
for page in doc_hdfc:
    full_text_hdfc += page.get_text()

chunks_hdfc = text_splitter.split_text(full_text_hdfc)
metadata_hdfc = parse_filename(filepath_hdfc)

documents_hdfc = []
for chunk in chunks_hdfc:
    chunk_metadata = metadata_hdfc.copy()
    chunk_metadata["section"] = "unclassified"
    documents_hdfc.append({"text": chunk, "metadata": chunk_metadata})

texts_hdfc = [doc["text"] for doc in documents_hdfc]
metadatas_hdfc = [doc["metadata"] for doc in documents_hdfc]
ids_hdfc = [f"{m['company']}_{m['quarter']}_{m['year']}_chunk{i}" for i, m in enumerate(metadatas_hdfc)]

embedded_vectors_hdfc = embeddings.embed_documents(texts_hdfc)

collection.add(
    documents=texts_hdfc,
    embeddings=embedded_vectors_hdfc,
    metadatas=metadatas_hdfc,
    ids=ids_hdfc
)

print(f"Collection now has {collection.count()} total documents")
```

```python
query_text = "credit risk"
query_embedding = embeddings.embed_query(query_text)

results = collection.query(
    query_embeddings=[query_embedding],
    n_results=5,
    where={"market": "India"}
)

for i, (doc, meta) in enumerate(zip(results["documents"][0], results["metadatas"][0])):
    print(f"--- Result {i} ---")
    print(f"Company: {meta['company']}, Market: {meta['market']}, Quarter: {meta['quarter']}")
    print(doc[:200])
    print()
```
